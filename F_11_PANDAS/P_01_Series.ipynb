### Creating a Series in Multiple Ways
#### 1. List
#### 2. Numpy Array
#### 3. Dictionary
#### 4. Custom Index
#### 5. Scalar Value

```python
## Series form list
country = ['India','Chine','America','Pakistan']
pd.Series(country)
```

```python
batsman_runs=[4,12,34,23,64,34,67]
pd.Series(batsman_runs,dtype='int32')
```

```python
subject_name = ['Maths','Physics','Chemistry','Biology','English']
subject_marks = [48,94,76,90,96]

pd.Series(data=subject_marks,index=subject_name,dtype='int32')
```

```python
## creating a series using a Numpy array
values = np.arange(10,21)
pd.Series(values)
```

### Dictionary to Series
