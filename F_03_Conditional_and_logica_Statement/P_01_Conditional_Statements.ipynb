## Nested Conditional Statement inside if statement inside another if statement

```python
# Compare the height of three friends

Amit_height=int(input("Enter that what your height is Amit: "))
Prabhat_height=int(input("Enter that what your height is Prabhat: "))
Shivy_height=int(input("Enter that what your height is Shivy: "))

if Amit_height>Prabhat_height:
    if Prabhat_height>Shivy_height or Amit_height>Shivy_height:
        print("Amit is the tallest guy")

elif Prabhat_height>Amit_height:
    if Prabhat_height>Shivy_height or Amit_height>Shivy_height:
        print("Prabhat is the tallest among the three friends")

else:
    print("Shivy is the tallest among the three friends")
```
