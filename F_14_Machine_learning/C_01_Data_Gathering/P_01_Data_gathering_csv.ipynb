# Data Gathering Using read_csv() ---> CSV File


### 1. How to Read content from a csv File

```python
import pandas as pd
```

```python
pd.read_csv('titanic.csv')
```

```python
# for reading tsv(tab separated value) dataset
## pd.read_csv('data_set_name.tsv',set="\t")
```

#### read_csv() --> 1. sep (sepration) --> how values are separated from each other in a given file 

##### csv --> neel,24 --> sep=","
##### tsv --> neel\t24 --> sep="\t"

### 2. names --> names is going to take a list of names that we can provide to our column names 
#### ---> names=['Name','Age',....]

```python
pd.read_csv('titanic.csv').head()
```

```python
pd.read_csv('titanic.csv',names=['P_Id','Pclass','P_Name','Gender','Age','Sibling or Spouce','Parch','P_Ticket','P_Fare','P_Cabin','P_Embarked']).head()
```

### 3. header --> header parameter makes sure that any of the row name will be shown as the column name only based on indexing of rows

```python
pd.read_csv('titanic.csv',header=0).head()
```

### 4. index_col --> keep one of our column as the index

```python
pd.read_csv('titanic.csv',index_col='PassengerId').head()
```

### 5. usecols --> what columns we should use to show out of all of the columns we are having in the dataFrame --> use a list to mention all of the column names which are required 

```python
pd.read_csv('Student Depression Dataset.csv',usecols=['id','Gender','Age','City','Profession','Academic Pressure','CGPA','Study Satisfaction','Job Satisfaction','Sleep Duration','Dietary Habits','Degree','Have you ever had suicidal thoughts ?','Depression']).head()
```

#### squeeze --> DataFrame ---> Series --> squeeze --> True --> Combination of usecols and squeeze --> True

```python
pd.read_csv('titanic.csv',usecols=['Name']).squeeze()
```

### 6. nrows --> it is going to help us with a fixed number of rows out of the complete data --> 100 --> only 100 rows out of all rows will be shown

```python
pd.read_csv('Student Depression Dataset.csv')
```

```python
pd.read_csv('Student Depression Dataset.csv',nrows=1000)
```

### 7. encoding --> It makes sure that the data we are reading is in utf-8 format --> encoding ='latin-1'

```python
## Encoding issue --> utf-8 --> Manually keep the encoding --> utf-8 --> Read the content from that dataset
```

```python
pd.read_csv('titanic.csv',encoding='latin-1').head()
```

```python
pd.read_csv('titanic.csv').head()
```

### 8. on_bad_lines ---> we can have some rows with 8 values while others will have 9 values so these rows with 8 values are required to be removed --> on_bad_lines = 'skip' or 'warn'

```python
pd.read_csv('titanic.csv',on_bad_lines='skip')
```

```python
pd.__version__
```

### 9. dtype --> it can assign initially only with a specific data type to a column --> dtype=['col_1','col2':float]

```python
pd.read_csv('titanic.csv',dtype={'Age':'float32'})
```

```python
pd.read_csv('Student Depression Dataset.csv',dtype={'Age':int,'Work Pressure':int,'Academic Pressure':int})
```

### 10. .info()

```python
pd.read_csv('Chocolate Sales.csv').info()
```

```python
pd.read_csv('Chocolate Sales.csv')
```

```python
pd.read_csv('Chocolate Sales.csv',parse_dates=['Date'])
```

```python
pd.read_csv('Chocolate Sales.csv',parse_dates=['Date']).info()
```

### 11. na_values --> can help us out with mentioning what values to be considered as a missing value

#### na_values=['',':','No']

```python
pd.read_csv('titanic.csv',na_values=["NaN"])
```
