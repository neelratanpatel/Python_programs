```python
m=sales[sales['Product']=='Mint Chip Choco']
d=sales[sales['Product']=='85% Dark Bars']
p=sales[sales['Product']=='Peanut Butter Cubes']
```

```python
m=m.groupby('Date')['Amount'].sum()
d=d.groupby('Date')['Amount'].sum()
p=p.groupby('Date')['Amount'].sum()
```

```python
m.index=pd.CategoricalIndex(m.index,categories=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],ordered=True)
d.index=pd.CategoricalIndex(d.index,categories=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],ordered=True)
p.index=pd.CategoricalIndex(p.index,categories=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'],ordered=True)
```

```python
m=m.sort_index()
d=d.sort_index()
p=p.sort_index()
```

```python
plt.Figure(figsize=(6,4))
plt.xlabel('Month Name')
plt.ylabel('Total Sales')
plt.bar(m.index,m.values,color='#fcba03',label='Mint Chip Choco')
plt.bar(d.index,d.values,color='#03fc41',label='85% Dark Bars')
plt.bar(p.index,p.values,color='#df03fc',label='Peanut Butter Cubes')
plt.legend()
plt.show()
```
