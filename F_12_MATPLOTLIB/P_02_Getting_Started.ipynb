### Stacked Bar Chart
#### Instead of Keeping the Bars in Side by Side Direction, we can keep the Bars in the Stacked Manner i.e, One On Top of other and in order to Figure out which category is having Maximum Sale, we can have a look over the Length of hte Bar

#### Stacked bar Chart --> bar() --> parameter --> bottom --> this bottom parameter specific that what bar i want to keep below my current bar 

#### Red Bar
#### Green Bar --> 
#### Blue Bar --> bottom --> Green Bar + Red Bar

```python
plt.Figure(figsize=(6,4))
plt.xlabel('Month Name')
plt.ylabel('Total Sales')
plt.bar(m.index,m.values,color='#fcba03',label='Mint Chip Choco')
plt.bar(d.index,d.values,color='#03fc41',label='85% Dark Bars',bottom=m.values)
plt.bar(p.index,p.values,color='#df03fc',label='Peanut Butter Cubes',bottom=m.values+d.values)
plt.legend()
plt.show()
```
