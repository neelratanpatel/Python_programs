## Typrs of Operators
### 1. Arithmetic Operators
####    a. Addition (+)
####    b. Substraction (-)
####    c. Product or Multiplication (*)
####    d. Division (/)
####    e. Modulus (%)
####    f. Floor (//)
####    g. Exponent (a**b)
### 2. Comparison Operators
####    a. Greater than (>)
####    b. Less than (<)
####    c. Greater than or equal to (>=)
####    d. Less than or equal to (<=)
####    e. Equality Check (==)
####    f. Not equal to (!=)
### 3. Logical Operators
####    a. and --> Whether if both conditions are true or not
####    b. or --> Whether if any of the operation is true or not
####    c. not --> Negation or to inverse the result of the given operation
