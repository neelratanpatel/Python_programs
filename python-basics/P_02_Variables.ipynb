## Input()

```python
print(Name)
```

### Input function only takes the input in a string format until and unless we are not going to explicitly take the input in integer or other format

```python
Age=input("Enter you Age: ")
```

```python
print(type(Age))
print(Age)
```

```python
Age=float(input("Enter Your Age : "))
print(Age)
print(type(Age))
Age = int(Age)
print(Age)
print(type(Age))
```

### Building a Simple Calculator

```python
value_1 = int(input("Enter your first Value: "))
value_2 = int(input("Enter your second Value: "))
print(value_1," + ",value_2," = ",value_1+value_2)
print(value_1," - ",value_2," = ",value_1-value_2)
print(value_1," * ",value_2," = ",value_1*value_2)
print(value_1," / ",value_2," = ",value_1/value_2)
print(value_1," % ",value_2," = ",value_1%value_2)
print(value_1," // ",value_2," = ",value_1//value_2)
print(value_1," ** ",value_2," = ",value_1**value_2)
```
