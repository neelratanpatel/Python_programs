### Types of Constructors
#### 1. Default Constructor
#### 2. Parameterized Constructor

### 1. Default Constructor 
#### Is doesnt have the parameters inside it and it ia already inside it and it is only having the self keyword inside it. It assigns a default value to all of the objects.

```python
class Dog:
    def __init__(self):
        self.name="Labrador"
        self.color="Brown"

    def show_info(self):
        print(f"This dog is {self.name} and the color of hte dog is {self.color}")
```

```python
dog1 = Dog()
```

```python
dog1.show_info()
```

### 2. Parameterized Constructor
#### It consists of parameters inside it so that every object can have a specific attribute associated to it .
#### It allows us to pass aruments while creating an object.

```python
class Dog:

    def __init__(self,bread,color):
        self.bread = bread
        self.color = color

    def show_info(self):
        print(f"the dog is {self.bread} and is having color is {self.color}")
```

```python
dog2 = Dog("Golden Retriever","Golden")
```

```python
dog2.show_info()
```

```python
dog3 = Dog("Husky","White")
```

```python
dog3.show_info()
```

## Methods
#### functionality that a specific object is going to have 

### Types of Methods
### 1. Instance Method
#### Instance Methods operate on Object Attributes.
#### The Instance Method is going to use self keyword to access and modify object attributes.
#### These Methods are called using an object. 



```python
class Avengers:
    def __init__(self,name,power):
        self.name=name
        self.power=power
    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}")
```

```python
Iron_man=Avengers("Iron man","Genius Intellect")
```

```python
Iron_man.show_details()
```


### 2. Class Method
#### Is the method of ht ewhole class or the whole blueprint which will remain the same for all of the objects than such kind of Method is being reffered as a Class Method.
#### In class Method, These Methods wordk with the class level attributes.
#### They are going to use cls instead of self
#### They are called using the class name as well


```python
class Avengers:

    origin='Avenger'

    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    def self_info(self):
        print(f"My name is {self.name} and my power is {self.power}")
    
```

```python
Iron_man=Avengers('Iron Man','Genius Intellect')
```

```python
Iron_man.power
```

```python
Iron_man.self_info()
```

```python
Iron_man.origin_info()
```

```python
Thor=Avengers('Thor','God of Thunder')
```

```python
Thor.self_info()
```

```python
Thor.origin_info()
```

### 3. Static Method
#### Static Methods do not use self or cls.
#### They work like normal functions but belong to class.
#### They are going to be designed by a abbraviation #staticmethod
#### static method can also be called by the classname also

```python
class Avengers:
    @staticmethod
    def mission():
        print("All Avengers must protect the America")

    origin="Avenger"

    def __init__(self,name,power):
        self.name=name
        self.power=power

    @classmethod
    def origin_info(cls):
        print(f"The ogigin for the Avenger is :{cls.origin}")

    def introduction(self):
        print(f"My name is {self.name} and power is that i have {self.power}")
```

```python
Iron_man = Avengers('Iron Man','Genius intellect')
```

```python
Iron_man.introduction()
```

```python
Iron_man.origin_info()
```

```python
Iron_man.mission()
```
