# Object Oriented Programming

#### In Object Oriented Programming we are trying to match up with the real world scenario by using Programming.

#### In OOPs we are going to learn about classes and objects

#### By a class we mean to provide a bluepring that must remain the same for each and every single object for us

#### These Objects can have different functionalities from each other and different characteristics from each other as well but they must follow the similar blueprint

#### By a blueprint or a class what we mean to have a means of providing the functionalities and the characteristics to our objects

### class Class_name:
#### def __init__(self,name,power,weapon,nationality):
#### self.name=name
#### self.power=power
#### self.weapon=weapon
#### self.nationality=nationality

#### def introduction(self):
#### print(f"Hello my name is {self.name}")

```python
class Avenger:
    def __init__(self,name,power,weapon,nationality='American'):
        self.name = name
        self.power = power
        self.weapon = weapon
        self.nationality = nationality
    def introduction(self):
        print(f"Hello , my name is {self.name}")

    def carrying_weapon(self):
        print(f"i am carrying the weapon {self.weapon}")
```

```python
Iron_man = Avenger("Iron Man","Genius Intellect and Technology","Repulsor beams")
```

```python
Iron_man.introduction()
Iron_man.carrying_weapon()
```

```python
class Cartoon:
    def __init__(self,name,show,catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase

    def introduction(self):
        print(f"Hello my name is {self.name}")

    def speak(self):
        print(f"{self.name} says {self.catchphrase}")
```

```python
Michey_mouse = Cartoon("Michey Mouse","Michey Mouse clubhouse","Oh boy !")
```

```python
Michey_mouse.show
```

```python
Michey_mouse.introduction()
```

```python
Michey_mouse.speak()
```

```python
Spongebob = Cartoon("Spongebob Squarepants","Spongebob Squarepants","I'm ready !")
```

```python
Spongebob.speak()
```

```python
Tom = Cartoon('Tom','Tom and jerry','..... silent but expressive')
```

```python
Tom.speak()
```
