## 3. Encapsulation

#### Encapsulation is a way of wrapping data (Variables and Methods) inside a capsule or a class as a Single Unit.
#### It restricts the direct access to some of the objects component's which is a means of preventing accidental inference and misuse of the data.

### Key concepts of Encapsulation:
#### 1. Public Members
#### 2. Private Members
#### 3. Protected Members

#### getter and setter methods

### 1. Public Members
#### Accessible from Anywhere or Attributes are accessible outside of the class and these members are accessible by the derived class
#### public member attrbute is directly declared


```python
class IronMan:
    def __init__(self,name,power,weapon):
        self.name=name
        self.power=power
        self.weapon=weapon

    def introduction(self):
        print(f"Hello my name is {self.name} and i have {self.weapon}")
```

```python
## dir() --> it takes the input as the object name and gives us this idea that which member is public or protected or private
```

```python
ironman = IronMan('Neel','Software Engineer','java,python')
```

```python
dir(ironman)
```

```python
ironman.name
```

```python
ironman.power
```

```python
ironman.weapon
```

```python
ironman.introduction()
```

### 2. Private Members 
#### A Private member is not accessible outside the class and not even in the derived class and in order to have the access over these private members or to bring some modification on these members we use getter and setter method.
#### Private Members are declared by usgin double underscore '__methodname' before the method name



```python
class Avenger:
    def __init__(self,name, phone_number):
        self.name = name
        self.__phone_number=phone_number

    def information(self):
        print(f"my name is {self.name} and my phone number is : {self.__phone_number}")
```

```python
avenger = Avenger('Neel','123456')
```

```python
avenger.name
```

```python
avenger.phone_number
```

```python
avenger.information()
```

### 3. Protected Members
#### Protected members are the data which is not accessible outside the class but the protected data is going to be accessible by the derived class or child class 

#### for having the protected member you can mention your protected member as '_attribute'or by using before a single underscore 

```python
class GrandParent:
    def __init__(self,name,locker_key):
        self.name=name
        self._locker_key=locker_key

    # getter method
    def locker_info(self):
        print(f"the locker key info is {self._locker_key}")
```

```python
class Parent(GrandParent):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveals(self):
        print(f"The locker key information is : {self._locker_key}")
        
```

```python
grandpa = GrandParent('neel','123456')
```

```python
grandpa.name
```

```python
grandpa.locker_key
```

```python
parent = Parent('neel','123456')
```

```python
parent.locker_key
```

```python
parent.child_reveals()
```

### getter Method
#### it will be helping us out with accessing the private data

```python
class Captain:
    def __init__(self,name,address):
        self.name=name
        self.__address = address

    #getter method
    def accessing_add(self):
        print(f"My address is {self.__address}")
```

```python
captain = Captain('Neel Ratan','Varanasi')
```

```python
dir(captain)
```

```python
captain.address
```

```python
captain.accessing_add()
```

### Setter Method
#### Setter Method is being used for modifying our private data

```python
class Animal:
    def __init__(self,name,sound):
        self.name=name
        self.__sound=sound

    def sound_info(self):
        print(f"The animal is going to sound as : {self.__sound}")

    def sound_modify(self,new_sound):
        self.__sound=new_sound
        print(f"New sound for the Animal is : {self.__sound}")
        
```

```python
animal = Animal('Bruno','Woof Woof!')
```

```python
animal.name
```

```python
animal.sound
```

```python
animal.sound_info()
```

```python
animal.sound_modify('meaw meaw')
```

```python
animal.name
```

```python
animal.sound_info()
```

### Bank Account System

```python
class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.__balance=int(balance)

    def get_balance(self):
        return (f"The Total balance is : {self.__balance}")

    def deposit(self,ammount):
        if int(ammount)>0:
            self.__balance+=ammount
            print( f" {self.get_balance()} \n Deposit Ammount is : {ammount}")
        else:
            print(f"Deposited Ammount is must be greater than zero : {ammount}")
```

```python
bandAccount = BankAccount('neel','5000')
```

```python
bandAccount.get_balance()
```

```python
bandAccount.deposit(-500)
```
