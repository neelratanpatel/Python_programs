## 3. Encapsulation

#### Encapsulation is a way of wrapping data (Variables and Methods) inside a capsule or a class as a Single Unit.
#### It restricts the direct access to some of the objects component's which is a means of preventing accidental inference and misuse of the data.

### Key concepts of Encapsulation:
#### 1. Public Members
#### 2. Protected Members
#### 3. Private Members
#### getter and setter methods
