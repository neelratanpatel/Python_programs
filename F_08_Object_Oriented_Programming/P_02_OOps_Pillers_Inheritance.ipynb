## 1. Inheritance
#### Inheritance is a fundamental concept in Object Oriented Programming because of which the child class is able to inherit the properties Or Attributes and functionalities of the parent class 
#### Just like a child is going to inherit the properties from parent and parent is going to inherit the properties from the gradparent and definitiely in this hierarchy the grandchild can have some features different from the parent and grandparent as well.

### In python typically we get to use 5 Types of Inheritance:
#### 1. Single Inheritance
#### 2. Multilevel Inheritance
#### 3. Multiple Inheritance
#### 4. Hierarchical Inheritance
#### 5. Hybrid Inheritance

### 1. Single Inheritance
#### The features or the Attributes from one parent class will be getting inherited into one child class.
#### Parent Class(Properties --> Attributes and Functionalities)
####            ^
####            |
####            |
#### Chaild Class

```python
class Avenger:
    def __init__(self,name,power):
        self.name=name
        self.power=power
    def introduction(self):
        print(f"Hello My name is {self.name}")

class Ironman(Avenger):
    def __init__(self, name, power, weapon):
        super().__init__(name, power)
        self.weapon=weapon
    def weapon_details(self):
        print(f"I use {self.weapon}")
```

```python
iron = Ironman('ironman','Genius intellect','Repulsor beams')
```

```python
iron.name
```

```python
iron.weapon
```

```python
iron.weapon_details()
```

```python
iron.introduction()
```

### 2. Multilevel Inheritance
#### The properties or methods and attributes of the grandfather class can be inherited by the parent class and than the properties of the parent class will be inherited by the child class. In a proper heirarchy the inheritance process is going to work for us.

#### Grandparent (properties)
#### |
#### |
#### Parent (Properties)
#### | 
#### |
#### Child

```python
class Grandparent:
    def __init__(self,nationality,boldness,anger_issues):
        self.nationality = nationality
        self.boldness = boldness
        self.anger_issues = anger_issues

    def intoduction(self):
        print(f"Hello my nationality is : {self.nationality} and about my anger_issues if you want to ask about than the answer is : {self.anger_issues}")
```

```python
class Parent(Grandparent):

    def __init__(self,nationality,boldness,anger_issues,intelligence):
        super().__init__(nationality,boldness,anger_issues)
        self.intelligence = intelligence

    def intellifence_details(self):
        print(f"About my intelligence if you guys want to know that whether if i am intelligent or not then the answer is : {self.intelligence}")
```

```python
class Child(Parent):
    def __init__(self,nationality,boldness,anger_issues,intelligence,charm):
        super().__init__(nationality,boldness,anger_issues,intelligence)
        self.charm=charm

    def charm_details(self):
        print(f"About my charm if you guys want to know that whether if i an charming or not then the answer is : {self.charm}")
```

```python
parent = Parent("Indian",'yes','No','yes')
```

```python
parent.intoduction()
```

```python
child = Child('Indian','No','Yes','Yes','Yes')
```

```python
child.intoduction()
```

```python
child.intellifence_details()
```

```python
child.charm_details()
```

```python
child.nationality='american'
```

```python
child.intoduction()
```

### 3. Multiple Inheritance
#### In Multiple Inheritance, The properties of the parent class or mltiple parent class will be getting inherited into the child class

#### Parent_1 (Properties) + Parent_2 (Properties) + Parent_3 (Properties)
#### |
#### |
#### Child

```python
class IronMan:
    def __init__(self,name,power,suit,weapon,nationality="American"):
        self.name=name
        self.power=power
        self.suit=suit
        self.weapon=weapon
        self.nationality=nationality

    def introduction(self):
        print(f"My name is {self.name} and powers are : {self.power}")

class CaptainAmerica:
    def __init__(self,name,power,suit,weapon,nationality='American'):
        self.name=name
        self.power=power
        self.suit=suit
        self.weapon=weapon
        self.nationality=nationality

    def weapon_details(self):
        print(f"I carry the weapon as : {self.weapon}")
```

```python
class Hulk(IronMan,CaptainAmerica):
    def __init__(self,name,power,suit,weapon,anger_issues,nationality='American'):
        super().__init__(name,power,suit,weapon,nationality)
        self.anger_issues=anger_issues

    def anger_description(self):
        print(f"About my anger issues if i am having or not then the answer is : {self.anger_issues}")
```

```python
hulk = Hulk('Hulk','amazing power and Strength','Mark 85 Iron Man suit','Repulsor beams and shield','Yes','American')
```

```python
hulk.introduction()
```

```python
hulk.anger_description()
```

```python
hulk.weapon_details()
```
