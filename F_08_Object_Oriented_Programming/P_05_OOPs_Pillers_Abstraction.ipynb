# Abstraction

#### Abstraction is the concept of Hiding the implementation details and only showing the neccessary features of the object.
#### This helps in reducing programming complexity and efforts.
#### Abstraction allows us to work with an object with what an object does rather than how it does it.

## How we can Achieve Abstraction in Python
#### 1. Abstract class
#### 2. Abstract method


## 1. Abstract class
#### A class that cannot be in initiated or that menas for an abstract class we cant create the object of this class.

## 2. Abstract method 
#### Abstract method just because ot cant accessed by the object of the abstract class itself thats why we have to call this abstract method in our child class and the implementation of the abstract method is not provided inside the abstract class but must be provided in the child class.

### The Abstract method is being defined by @abstractmethod

```python
from abc import ABC,abstractmethod

class Vechile(ABC):

    @abstractmethod
    def start(self):
        pass
    
    @abstractmethod
    def stop(self):
        pass

class Car(Vechile):
    def start(self):
        print("The car starts with an ignition key")
    
    def stop(self):
        print("The car stops with break padals")

class ElectronicScooter(Vechile):
    def start(self):
        print("Electronic Scooter is going to start with a power button")
    
    def stop(self):
        print("Electronic Scooter stops with an electronic break")
```

```python
car=Car()
```

```python
car.start()
```

```python
scooter.start()
```
