## 2. Polymorphism

#### Polymorphism means MANY FORMS in Greak.
#### In OOPs , Polymorphism allows objects of Different classes as if they are the object of the same class & it enables reusability and makes programs more flexible.

### Types of Polymorphism:
#### 1. Method Overriding (Runtime Polymorphism)
#### 2. Method Overloading(Compile Time Polymorphism)
#### 3. Operator Overloading

### 1. Method Overriding (Runtime Polymorphism)
#### A child class provides a specific implementation of a method that has already been defied into the parent class. This allows Dynamic Selection at Runtime.

```python
class GrandParent:
    def speak(self):
        print("Hello My name is Neel from GrandParent")

class Parent(GrandParent):
    def speak(self):
        print("Hello My name is Neel from Parent")

class Child(GrandParent):
    def speak(self):
        print("Hello my name is Neel from Child")
```

```python
def attribute_provider(objects):
    return(objects.speak())
```

```python
grand = GrandParent()
```

```python
attribute_provider(grand)
```

```python
parent = Parent()
```

```python
attribute_provider(parent)
```

```python
child = Child()
```

```python
attribute_provider(child)
```

```python
class Shape:
    def area(self):
        return "This method returns the area of the shape"
```

```python
class Ractangle(Shape):
    def __init__(self,width,height):
        self.width=width
        self.height=height
    
    def area(self):
        return self.width*self.height
```

```python
class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius

    def area(self):
        return 3.14*self.radius*self.radius
```

```python
shape=Shape()
```

```python
shape.area()
```

```python
ractangle = Ractangle(4,4)
ractangle.area()
```

```python
circle=Circle(5)
circle.area()
```
