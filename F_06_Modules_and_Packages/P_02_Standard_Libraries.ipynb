#### 3. csv library --> Most of the time we are going to work with csv files so thats why knowledge with csv library functionality matters

```python
import csv
```

```python
with open('example.txt', mode='w', newline='' ) as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Neel',23])
```

```python
with open('example.txt', mode='r') as file:
    render = csv.reader(file)
    for row in render:
        print(row)
```

#### 7. JSON Library


```python
# When we are going to work with Nosql database data then the data inside the Nosql Database is going to be their in the Json format and along with that when we are going to work with the database as well then again these dataset as will then again these dataset will be consisting of data is the json formate itself
```

```python
# When we get the Json data , we get in a json string format so by using json library we can convert our string dta into dictionary format for serializing our data
```

```python
import json
```

```python
data = {'Name':'Neel', 'Age':23}
json_str = json.dumps(data)
print(json_str)
print(type(json_str))
```

```python
json_dict = json.loads(json_str)
print(json_dict)
print(type(json_dict))
```
