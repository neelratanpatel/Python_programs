# NUMPY

#### Numpy is a Fundamental Package for scientific camputing in python. It is a Python Llibrary that provides a Multi Dimenstional Arry object and various derived objects such as Masked Arrays or Matrices and an assortment of routines for fast operations on Arrays including Mathematical, Logical, Shape Manipulation, Sorting, Selecting, I/O, Discrete Fourier Transform, Basic Linear Algebra, Statical Operations, Random Simulation etc.

####  At the core of the Numpy Package is the n dimentional Array object.

### Why Numpy ? 
#### 2000 - 2009, Data science --> Matlab, R Language
#### Data types in Python --> List, Dictionary --> Slow Syntax --> Python --> Easy
#### We needed Speed and less Memory to work with the hug ammount of data.
#### Numpy --> c language --> wrapped --> Python language 
#### Numpy --> syntax --> Easy for beginners --> performing scientfic calculations --> Creating Multi Dimentional Arrays --> Speed is Faster for applying Operations --> Less memory
#### Libraries like Scikit learn, Tensorflow and pytorch require input data in Numpy Arrays or Tensors.

### Numpy Arrays Vs Python Sequences
#### Numpy Arrays have a fixed size at creation, unlike Python Lists which grew dynamically, Channing the size of an n dimetional array will create a new arry and delete the original.
#### The elements in a Numpy Array are required to be of the same data type and thus will be of the same size in memory .
#### Numpy Arrays facilitats advanced Mathemetical and other types of operations on large numer of data. Typically, such operations are executed more effeiciently with less code than is possible using Python's built in sequances.

## Stacking and Spliting

#### Stacking --> Joining The Arrays Together
#### Horizontal Stacking --> Now by Row joining is done --> hstack()
#### Vertical Stacking --> Column by Column Joining is done --> vstack()
#### Horizontal split --> np.hsplit(array,number of parts you want your array to be splitted at)
#### vertical split --> np.vsplit(array,number of parts)

```python
import numpy as np
```

```python
array_one = np.arange(1,13).reshape(3,4)
```

```python
array_one
```

```python
array_two = np.arange(13,25).reshape(3,4)
```

```python
array_two
```

#### Horizontal Stacking --> Now by Row joining is done --> hstack()

```python
horizontal_array = np.hstack((array_one,array_two))
```

```python
horizontal_array
```

```python
horizontal_array = np.hstack((array_one,array_two,array_one,array_two))
```

```python
horizontal_array
```

#### Vertical Stacking --> Column by Column Joining is done --> vstack()

```python
vertical_array = np.vstack((array_one,array_two))
```

```python
vertical_array
```

#### Horizontal split --> np.hsplit(array,number of parts you want your array to be splitted at)

```python
np.hsplit(vertical_array,2)
```

#### vertical split --> np.vsplit(array,number of parts)

```python
np.vsplit(vertical_array,2)
```

```python
vertical_array
```

```python
np.vsplit(vertical_array,3)
```
