#### tuples inside nested list

```python
nested_tuple=[(1,2,3),(4,5,6),(11,1,2,1,3,25,56)]
```

```python
nested_tuple[0][:]
```

#### iterating over a tuple inside a tuple

```python
for i in nested_tuple:
    for j in i:
        print(j, end=' ')
    print()
```
