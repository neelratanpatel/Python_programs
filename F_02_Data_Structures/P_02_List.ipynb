### Syntax for List comprehension it
#### [expression for item in iterable if condition]
#### expression --> what we want to print
#### for item in iterable --> looping statement
#### if condition --> conditional statement

```python
# I nee to store values from 1 to 50 inside the list
```

```python
list_item=[x for x in range(1,51)]
```

```python
list_item
```

```python
list_squared = [items**2 for items in range (1,16)]
```

```python
list_squared
```

```python
# if i am only interested in looking over the square values which are even
even_items=[x**2 for x in range(1,19) if x%2==0]
```

```python
even_items
```

## Nested List Comprehension
#### we will be having the combination of multiple lists in our single list itself

```python
list_items
```

```python
pair = [[i,j] for i in list_items for j in length_list]
```

```python
lst_1 = [1,3,4]
lst_2 = [2,6,9]

paired_list = [[i,j] for i in lst_1 for j in lst_2]
```
