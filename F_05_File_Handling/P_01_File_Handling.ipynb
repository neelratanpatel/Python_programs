# File Handling
### 1. Read The content from My text file in the system


```python
## open() --> opening your file
## first Parameter --> File_name
## second Parameter --> Mode
## Mode --> It decides the way in which we are working with this file

# Mode --> r --> Reading the content from the file
# Mode --> w --> Write the content into the file
# Mode --> r+ --> Read the content first and then writing the content into the File
# Mode --> w+ --> Writing the content first and then reading from it 
```

```python
## Reading your File
## read() -->
```

```python
f=open('example.txt','r')
content = f.read()
print(content)
f.close()
```

### 2. Writing the content into the file

```python
## write () --> it overwrites the content into the text file
## appent () --> it adds the data from the end of the file
## Mode --> w --> content wii be overwritten
## Mode --> a --> content will be added from the end
```

```python
# write () --> helps us out in writing the content into the file

f=open('example.txt','w')
f.write('Hello We are now writing the content into our File')
f.close()
```

```python
# append() --> appending the content into our file

f=open('example.txt','a')
f.write('\nI am learning about Gen Ai Pre requisite Python')
f.close()
```

### 3. with --> It will be used for first of all opening the file aas a object and there is no requirement for closing the file as well

```python
with open('example.txt','r') as f:
    content_data = f.read()
    print(content_data)
```

### readlines() --> Read the line in a way there in it is like the lnes are stored inside a list {'Hello i am neel'.'am the ai learner','let learn python first'}

```python
# Reading the Total lines, Total words, Total Characters from a file
def count_lines(file):
    with open(file,'r') as files:
        lines = files.readlines()
        total_lines = len(lines)
        total_word = sum(len(line.split()) for line in lines)
        total_character = sum(len(line) for line in lines)
    return total_lines,total_word,total_character

file_path = 'Total_text.txt'
total_lines,total_word,total_character = count_lines(file_path)

print(f'Total lines -->  {total_lines} \nTotal words -->  {total_word} \nTotal character -->  {total_character}')
```

```python
## Write the content into a file and then after writing the content into that file we are going to read the content from that file as well --> seek()
```

### w+ --> first of all as W is their so it states that firs writing operations will be done and then Reading operations will be done
### seek() --> It is going to help us out with reading the content after writing from the starting only

```python
with open('Total_data','w+') as write_read_file:
    write_read_file.write("We are looking into W+ operations")
    write_read_file.write("\nIt means Writing and then Reading")
    write_read_file.seek(0)
    content_read_data = write_read_file.read()
    print(content_read_data)
```
