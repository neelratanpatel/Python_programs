# OS Library

### OS Library is able to interact with our Operating System and can let us work with Multiple Directories and Files

#### 1. getcwd() --> get current working directroy --> it is capable of prowiding us with the idea about the current Working Directories

```python
import os
```

```python
current_path = os.getcwd()
print('The current working directory is: ',current_path)
```

#### 2. mkdir() --> make directory --> It will be helping us out with creating the directory

```python
new_path = "P_02_Os_library_New_Folder"
os.mkdir(new_path)
print(f"The new directory {new_path} is created sucessfully")
```

#### 3. listdir() --> list directories --> Listing all of the directories present

```python
all_items = os.listdir('.')
print(all_items)
```

#### 4. join() --> joining multiple paths together

```python
## os.path.join() --> it will be joining multiple paths together

dir_name = "folder_data"
file_name = "file_data"
full_path=os.path.join(dir_name,file_name)
print(f'The full path is : {full_path}')
```

```python
## I want to join Multiple Path with the Current Path

current_complete_path = os.path.join(os.getcwd(),dir_name,file_name)
print(f'The full path is : {current_complete_path}')
```

#### 5. exists()

```python
## os.path.exists()

given_path="Gen Ai\\Practice_Gen_Ai\\F_05_File_Handling\\example.txt"

if os.path.exists(given_path):
    print("Yes this path exists")
else:
    print("This path does not exists")
```

#### 6. isfile()

```python
## os.path.isfile() --> This checks that whether if the current path is a file or not


import os

path = 'Total_text.txt'

if os.path.isfile(path):
    print("the current path belongs to a file")
elif os.path.isdir(path):
    print("the current path belongs to a directory")
else:
    print("the current path is neighther a file or a folder")
```

#### 7. isdir()

```python
## or.path.isdir() --> This checks that whether if the current path is a directory or not

path = 'hi'

if os.path.isfile(path):
    print("the current path belongs to a file")
elif os.path.isdir(path):
    print("the current path belongs to a directory")
else:
    print("the current path is neighther a file or a folder")
```

#### 8. abspath()

```python
## os.path.abspath() --> Absolute Path --> The complete path
## os.path.abspath() --> it provides us with the complete path of the absolute path of the file or directory
```

```python
file_name = 'Total_text.txt' # Total_text.txt --> Relative path
complete_path = os.path.abspath(file_name)

print(f"The complete path or absolute path for the {file_name} is: ", complete_path)
```
