# Streamlit

#### Streamlit is an Open source Python Library used to build Beautiful, interactive we applications for data science and machine learning with minimal effort.


### Easy to use 
#### 1. Fast Development
#### 2. Interactive UI Elements --> Buttons, Sliders, Text inpluts, Charts , .etc
#### 3. Great for data science --> Supports Pandas, Matlplotlib, Plotly and other libraries Deployment is easy

### how to start 
#### 1. install the streamlit 
##### pip install streamlit
#### 2. run streamlit file 
##### streamlit run app.py // app.py is the app_name.py
