## Recursive Function

#### going to call itself again and again

```python
# Factorial of a Number
# Factorial of 8 --> 8*7*6*5*4*3*2*1
# Factorial of 0 or Factorial of 1 --> 1
```

```python
def factorial(num):
    if num==0 or num==1:
        print(f"the factorial of {num} os : 1 " )
    else:
        factorial=1
        for i in range(1,num+1):
            factorial=factorial*i
        print(f"The factorial of num is : {factorial}")
        
```

```python
factorial(int(input("Enter the number: ")))
```

```python
def factorial_recursive(num):
    if(num==0 or num==1):
        return 1
    else:
        return num * factorial_recursive(num-1)
```

```python
factorial_recursive(5)
```

```python
# return num*factorial_recursive(num-1)
'''
base condition ---> which will stop the loop or which will let the function to step calling itself in our case, base condition was --> if i==0 or i==1
num* factorial_recursive(num-1)
num=5
5*factorial_recursive(4) & factorial_recursive(4) --> 4*factorial_recursive(3)
factorial_recursive(3) --> 3*factorial_recursive(2)
factorial_recursive(2) --> 2*factorial_recursive(1)
'''
```

```python
input_recursive=int(input("Enter the value: "))
print(f"the factorial value of {input_recursive} is : ",factorial_recursive(input_recursive))
```

### To create a function that converts temprature from celcius to fahrenheit or a fahrenheit to celcius

```python
## if temprature is in celcius, then celcius to fahrenheit conversion is 

'''
1 deg celcuus = 9/5 + 32
45 deg c = 45*9/5+32 f
'''

'''
1 fahrenheit celcius = 5/9 - 32
45 deg f = (45-32) * 5/9 c 
'''
```

```python
def temp_converter(temp_value,unit):
    if unit=='C':
        temprature = temp_value *9/5+32
        print(f"The temprature for {temp_value} deg Celcius is Fahrenheit in : ",str(temprature) + "F")
    elif unit=='F':
        temprature = round((temp_value - 32) * 5/9,2)
        print(f"The temprature for {temp_value} deg Fahrenheit in celcius is : ", str(temprature) +"C")
    else:
        print("Invalid value entered by the user side")
```

```python
temp_converter(2,'C')
```

```python
temp_converter(80,'F')
```

### To create a function that check the password streinght

```python
'''
Strong password -->
if your password:
    1. length is greater than 8 charactors
    2. password must consist of at least a lowercase charactor (a,b,c,d,..........)
    3. password must contain at least a Uppercase charactor (A,B,C,D,..........)
    4. password must contain at least a digit(1,2,3,4,5,...........)
    5. password must contain at least a special symbol (!,@,#,$,....)


Weak password --> 
    If a  password doesnt match any above candition then it will be considerd as a weak password
'''
```

#### islower() --> checks that whether if the current character is a lowercase character or not
#### isupper() --> checks that whether if the current character is a uppercase character or not

```python
def password_checker(password):
    if len(password)<8:
        print(f" the entered password : {password}  is a Weak password")
    elif not any (char.islower() for char in password):
        print(f" the entered password : {password}  is a Weak password")
    elif not any (char.isupper() for char in password):
        print(f" the entered password : {password}  is a Weak password")
    elif not any (char.isdigit() for char in password):
        print(f" the entered password : {password}  is a Weak password")
    elif not any (char for char in '!,@,#,$,%,^,&,*,(,),_'):
        print(f" the entered password : {password}  is a Weak password")
    else:
        print(f"The entered password {password} is a strong password ")
```

```python
password_checker('Neel@123')
```

### To create a function that check the String is palindrome String or not

```python
# Palindrom String
"""
Palindrom String is going to have charactors similar to it from both the sides or even if you will reverse the string still you will be getting the given string itself

palindrom --> aba, aabbaa,
not a palindrom --> neel, 

"""
```

```python
### Check that whether if the current string is a palindrom or not

def palindrom_checker(string):
    string_value = string.lower().replace(" ","")

    if string_value==string_value[::-1]:
        print ("The current string is a palindrome")
    else:
        print ("The current string is not a palindrome")
```

```python
palindrom_checker('Aba')
```

```python
palindrom_checker('bca')
```

## map()

#### Map function applies a given function operation to All items in an iterable or a List and returns a Map Object (An Iterator)

#### map() is going to help us out with avoiding unnecessary lines of code

```python
## Squaring all items of a list
## list --> [1,2,3,4,5,6]
## [] --> [1,4,9,16,25,36]

def square(list_items):
    empty_list=[]
    for item in list_items:
        squared_items = item**2
        empty_list.append(squared_items)
    print(empty_list)
```

```python
list_items = [1,2,3,4,5,6,7,8]
square(list_items)
```

```python
## similar problem statement with a Map()-------->

print(list(map(lambda x:x**2,[1,2,3,4,5,6,7,8])))
```

```python
## map() with basic function in it

def square_number(num):
    return num**2

print(list(map(square_number,[1,2,3,4,5,6,7,9])))
```

### Whenever you are going to pass the genera function inside your map function as the parameter then you dont have to use parenthesis
